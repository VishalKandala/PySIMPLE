```python
from pysimple import run
from pysimple.postprocess import plot_result
from pysimple.verification import poiseuille_metrics

completed = run('poiseuille', nx=8, ny=24)
print(f'converged: {completed.result.converged}')
print(f'iterations: {completed.result.iterations}')
print(poiseuille_metrics(completed.result, completed.grid, completed.case))
```

```python
plot_result('poiseuille.png', completed.result, completed.grid)
from IPython.display import Image, display
display(Image('poiseuille.png'))
```
